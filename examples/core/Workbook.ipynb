# Loading and processing a report

```python
import proteolyzer as pz

report = pz.read(r"report.parquet")
# report = pz.read(r"evidence.txt")  # MaxQuant output is recognized too
```

```python
report = report.process()
```

## What is in it

Counts are of distinct values, so a precursor seen in two channels of one run
counts once.

```python
report.summary()
```

## What it costs

`process()` narrows dtypes and replaces repeated strings with categories, which
on a real report takes a third to a half off the frame. This is where to look
if one is bigger than expected.

```python
report.memory().head(10)
```

## Pivoting to a quantitative matrix

```python
report.matrix('Ms1.Area', ['Precursor.Id'], ['Run']).matrix
```
